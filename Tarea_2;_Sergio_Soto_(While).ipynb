```python
#Ejercicio 1. Solicitar Valor de ph hasta Osciar entre 0-14:

#Establecimiento del "Contador"
pH = 15

#Establecimiento de los Requisitos de While
while pH < 0 or pH > 14:
  pH = float(input("Introduzca un valor para pH: "))
  if pH >= 0 and pH <= 14:
    break
  print("---------------------")
  print("Valor de pH inválido")
  print("---------------------")

#Resultado
print("----------------------------")
print(f"Valor de pH ({pH}) es válido")
```

```python
#Ejercicio 2. Conteo de un Valor de Átomos:

#Solicitud del Número de Átomos + Decoración
átomos = int(input("Introduzca el número de átomos: "))
print("---------")

#Establecimiento del "Contador"
x = 0

#Comienzo del Ciclo para Conteo de Átomos
while x < átomos:
  x = x + 1
  print(f"{x} átomos")
```

```python
#Ejericio 3. Acumulación Gradual de Masas hasta 100g

#Establecimiento del "Contador"
gramos = 0

#Comienzo del Ciclo para Sumatoria de Masas
while gramos < 100:
  suma = float(input("Ingrese una masa (g) para llegar al objetivo: "))
  gramos = gramos + suma

  if gramos >= 100:
    break #Establecimiento de Límite para ciertas Acciones

  print("-----------------------------")
  print(f"Masa acumulada: {gramos}g; sigamos")
  print("-----------------------------")

#Resultado
print("-----------------------------------")
print(f"Masa final: {gramos}g; objetivo: 100g")
```

```python
#Ejercicio 4. Número de Ciclos hasta Disminuir una Concentración hasta Valor Final:

#Establecimiento del "Contador"
concentración = 100
ciclos = 0

#Comienzo de Ciclo para Resta Porcentual
while concentración > 10:
  concentración = concentración * 0.80
  ciclos = ciclos + 1

#Resultado
print(f"Concentración final: {concentración:.2f} mol/L; número de ciclos necesarios: {ciclos}")
```

```python
#Ejercicio 5. Menú de Conversión Gramos-Kilogramos hasta Obtener Condicion de Salida:

#Establecimiento del "Contador"
opcion = ""

#Comienzo del Ciclo para Menú de Conversión o Salida
while opcion != "2":
  print("------------Menú de Conversión------------")
  print("------------------------------------------")
  print("1. Convertir Gramos(g) a Kilogramos (kg)")
  print("2. Salir del Menú de Conversión")
  print("------------------------------------------")

#Solicitud de Valor para Interacción con Menú (Donde Empieza el Proceso)
  opcion = input("Selecciones una opción (1-2) para continuar: ")

#Continuación/Interpretaciones del Ciclo
  if opcion == "1":
    print("------------------------------------------")
    gramos = float(input("Ingrese la cantidad de gramos (g) a convertir: "))
    kilogramos = gramos / 1000
    print("------------------------------------------")
    print(f"{gramos}g son equivalentes a {kilogramos}kg")
    print("------------------------------------------")
  elif opcion == "2":
    print("------------------------------------------")
    print("Saliendo del Menú de Conversión; muchas gracias...")
    print("------------------------------------------")
  else:
    print("------------------------------------------")
    print("Opción inválida; prueba con valores numericos disponibles")
    print("------------------------------------------")
```

```python
#Ejercicio 6. Demostración del Proceso de Enfriamiento de una Sustancia:

#Establecimiento del "Contador"
temperatura = 150

#Comienzo del Ciclo sobre el Enfriamiento de la Sustancia
while temperatura > 25:
  temperatura = temperatura - 10
  print("--------------------------")
  print(f"Temperatura actual: {temperatura}°C")

#Resultado
print("--------------------------")
print(f"Temperatura final: {temperatura}°C; alcanzada. Temperatura objetivo: 25°C")
```

```python
#Ejercicio 7. Solicitud de Masas Válidad (m > 0) y su Respectiva Sumatoria:

#Establecimiento de "Contadores"
ciclos = 0
masa_total = 0

#Comienzo del Ciclo para Solicitud y Sumatoria de Masas
while ciclos < 5:
  masa = float(input(f"Ingrese {ciclos + 1}° masa para el conteo/sumatoria: "))
  if masa > 0:
    masa_total = masa_total + masa
    ciclos = ciclos + 1

    if ciclos == 5:
      break #Para evitar mensajes innecesarios al final de la sumatoria

    print("----------------------------------------------------------")
    print(f"Valor ingresado válido; {ciclos}/5 ingresados, continue...")
    print("----------------------------------------------------------")
  else:
    print("----------------------------------------------------------")
    print("Valor ingresado inválido; ingresar valores superiores a 0")
    print("----------------------------------------------------------")

#Resultado
print("----------------------------------------------------------")
print(f"Ingreso y sumatoria de masas validado, valor final: {masa_total}")
print("----------------------------------------------------------")
```

```python
#Ejercicio 8. Neutralización de Ácido-Base entre 50mL de Disolución:

#Estableciminto del "Contador"
base = 0

#Comienzo del Ciclo para Adición de Base
while base < 50:
  base = base + 5

  if base == 50:
    break #Para evitar mensajes innecesarios

  print("---------------------------------------------------")
  print(f"Se llevan agregados {base}mL Base a los 50mL Ácidos")

#Resultado
print("---------------------------------------------------")
print(f"Se agregaron {base}mL de Base a los 50mL de Ácidos; se neutralizaron ambos pH")
```

```python
#Ejercicio 9. Reconocimiento de un Compuesto/Fórmula Químico (H2O) y Reiteración en la Solicitud:

#Establecimiento del "Contador"
compuesto = ""

#Comienzo del Ciclo hasta Reconocer Compuesto/Fórmula H2O
while compuesto != "H2O":
  compuesto = input("Adivina la fórmula química (Mayúsculas):")
  if compuesto != "H2O":
    print("----------------------------------------")
    print("Fórmula incorrecta, intenta de nuevo")
    print("----------------------------------------")

#Resultado
print("----------------------------------------")
print(f"Fórmula correcta, el cumpuesto era: {compuesto}")
```

```python
#Ejercicio 10. Determinación de Vidas Medias de una Muestra Radioactiva:

#Establecimiento de los "Contadores"
vidas_medias = 0
nucleos = 800

#Comienzo del Ciclo para Cálculo de Vida Media
while nucleos > 50:
  nucleos = nucleos / 2
  vidas_medias = vidas_medias +  1

#Resultados
print(f"Se requieren {vidas_medias} vidas medias para llegar a una cantidad de {nucleos} nucleos final")
```
