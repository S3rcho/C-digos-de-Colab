```python
import pandas as pd
import matplotlib.pyplot as plt
datos = pd.read_csv("casasboston.csv")

#Redefinir los nombres de las Columnas del CSV:
df = datos.copy()
df.columns = [
	"Ciudad",
	"Habitaciones",
	"Edad_casa",
	"Distancia_centro",
	"Impuesto",
	"Valor_mediano",]

#Mostrar datos del dataframe:
print ("Primeros Registros:")
print (df.head())
print ("----------------------------------------------")
print ("Informacion del dataframe:")
print (df.info())
print ("------------------------------------------------------------------------------")
print ("Resumen estatadisticos del dataframe:")
print (df.describe())
print ("------------------------------------------------------------------------------")

#Histogramas del "valor_mediano":
plt.figure(figsize=(10,6))
plt.hist(df["Valor_mediano"], bins=10, edgecolor="black")
plt.xlabel("Valor Mediano")
plt.ylabel("Frecuencia")
plt.title("Histograma del Valor Mediano de la Casa")
plt.show()

# Grafico de Dispersion "Valor Mediano vs Casa"
# Eje X "Habitaciones"
# Eje Y "Valores medianos"
plt.figure(figsize=(10, 6))
scatter = plt.scatter(
    df["Habitaciones"],
    df["Valor_mediano"],
    c=df["Valor_mediano"],
    cmap="viridis",
    alpha=0.5,)

plt.colorbar(scatter, label="Valor Mediano")
plt.title("Grafico de Dispersion de Habitaciones vs Valor mediano")
plt.xlabel("Numero de Habitaciones")
plt.ylabel("Valor Mediano")
plt.show()

#Para cambiar los ejes
     #plt.scatter(df["Valor mediano"], df["habitaciones"], alpha=0.5)
     #plt.show()
#Probar diferentes valores de transparencia
     #alpha=1
     #alpha=0.5
     #alpha=0.25
#Grafica Horizontal
#Ordenada por ciudad y valor mediano
#Mostrar los primeros 10 registros

top10 = (
    df.sort_values(by=["Valor_mediano"], ascending=False)
    .head(10))

plt.figure(figsize=(10,6))
plt.barh(top10["Ciudad"], top10["Valor_mediano"], color="green")
plt.title("Top 10 Ciudades por Valor mediano")
plt.xlabel("Valor Mediano")
plt.ylabel("Ciudad")
plt.show()

#Gráfica de Pie "Distribución de Ciudades"
conteo_ciudades = df["Ciudad"].value_counts()
plt.figure(figsize=(8, 8))
plt.pie(conteo_ciudades, labels=conteo_ciudades.index, autopct="%1.1f%%")
plt.legend(conteo_ciudades.index)
plt.title("Distribucion de Ciudades")
plt.show()
```
