```python
#Ejercicio 1. Clasificación de Valores de pH:

#Solicitud de su Valor de pH
pH = float(input("Ingrese el valor de pH (0-14): "))

#Reconocimiento/Clasificación del pH
if pH < 7 and pH >= 0:
  print("--------------------------------------")
  print("Su valor de pH se clasifica como Ácido")
elif pH > 7 and pH <=14:
  print ("-------------------------------------")
  print("Su valor de pH se clasifica como Base")
elif pH == 7:
  print("---------------------------------------")
  print("Su valor de pH se clasifica como Neutro")
else:
  print("----------------------------")
  print("Su valor de pH no es válido")
```

```python
#Ejercicio 2. Determinación del Estado Físico del Agua:

#Solicitud de la Temperatura del Agua
agua = float(input("Ingrese la temperatura del agua en grados Celsius: "))

#Reconocimiento/Clasificación del Estado del Agua
if agua < 0:
  print("-----------------------------")
  print("El agua esta en Estado Sólido")
elif agua > 0 and agua < 100:
  print("------------------------------")
  print("El agua esta en Estado Líquido")
elif agua > 100:
  print("------------------------------")
  print("El agua esta en Estado Gaseoso")
else:
  print("-------------------------------------")
  print("El agua esta Transicionando de Estado")
```

```python
#Ejercicio 3. Clasificación de Elementos en Metales, No Metales y Gases Nobles:

#Determinación de Elementos a Codificar
metales = ["Fe", "Cu", "Na", "Al", "Au", "Ag", "Ca", "K"]
no_metales = ["H", "C", "N", "O", "P", "S", "F", "Cl"]
gases_nobles = ["He", "Ne", "Ar", "Kr", "Xe", "Rn"]

#"Cartelera" de Elementos Codificados
print("Lista de Elementos:")
print("-------------------")
print("Fe", "Cu", "Na", "He", "Ne", "H", "C")
print("Ca", "Au", "Ar", "Kr", "N", "P", "S", "F")
print("Ag", "Al", "Cl", "Xe", "Rn", "K", "O")
print("-------------------")

#Solicitud de Elemento
elemento = input("Ingrese el símbolo del elemento: ")

#Reconocimiento/Clasificación del Elemento
if elemento in metales:
  print("-----------------------")
  print("El elemento es un Metal")
elif elemento in no_metales:
  print("--------------------------")
  print("El elemento es un No Metal")
elif elemento in gases_nobles:
  print("----------------------------")
  print("El elemento es un Gas Noble")
else:
  print("----------------------------------------------------")
  print("El elemento no se encuentra en la Lista de Elementos")
```

```python
#Ejercicio 4. Clasificación de Concentraciones Molares a Diluida, Moderada o Concentrada

#Solicitud de Concentración Molar
concentración = float(input("¿Cúal es su concentración molar (mol/L)?: "))

#Reconocimiento/Clasificación de su Concentración
if concentración < 0.1 and concentración >= 0:
  print("-------------------------------")
  print(f"{concentración} se clasifica como Diluida")
elif concentración <= 1 and concentración >= 0.1:
  print("-------------------------------")
  print(f"{concentración} se clasifica como Moderada")
elif concentración > 1:
  print("---------------------------------")
  print(f"{concentración} se clasifica como Concentrada")
else:
  print("-----------------------------------------")
  print(f"{concentración} no es un valor válido clasificable")
```

```python
#Ejercicio 5. Determinación de Riesgo de una Reacción según Temperatura:

#Solicitud de  los Valores de Temperatura
temperatura = float(input("Ingrese la temperatura (°C) de su reacción: "))

#Reconocimiento/Clasificación del Riesgo de la Reacción
if temperatura > -273.15 and temperatura < 40:
  print("-------------------------------------------------")
  print(f"{temperatura} °C vuelven a su reacción de riesgo: Estable")
elif temperatura >= 40 and temperatura <= 100:
  print("--------------------------------------------------")
  print(f"{temperatura} °C vuelven a su reacción de riesgo: Moderado")
elif temperatura > 100:
  print("----------------------------------------------")
  print(f"{temperatura} °C vuelven a su reacción de riesgo: Alto")
else:
  print("---------------------------------------------------------")
  print("La temperatura debe estar sobre Cero Absoluto (-273.15°C)")
```

```python
#Ejercicio 6. Reconocimiento del Elemento según su Número Atómico:

#Solicitud del Número Atómico
número_atómico = int(input("Ingrese un número atómico (1-5): "))

#Reconocimiento del Elemento Relacionado al Número Atómico
if número_atómico == 1:
  print("-------------------------------------")
  print(f"{número_atómico} corresponde al elemento: Hidrógeno")
elif número_atómico == 2:
  print("---------------------------------")
  print(f"{número_atómico} corresponde al elemento: Helio")
elif número_atómico == 3:
  print("--------------------------------")
  print(f"{número_atómico} corresponde al elemento: Litio")
elif número_atómico == 4:
  print("-----------------------------------")
  print(f"{número_atómico} corresponde al elemento: Berilio")
elif número_atómico == 5:
  print("--------------------------------")
  print(f"{número_atómico} corresponde al elemento: Boro")
else:
  print("-----------------------")
  print("Número atómico inválido")
```

```python
#Ejercicio 7. Comparación de Electronegatividades entre 2 Elementos:

#Solicitud de Ambas Electronegativadades
a = float(input("Ingrese la electronegatividad del primer elemento: "))
b = float(input("Ingrese la electronegatividad del segundo elemento: "))

#Reconocimiento/Comparación de Ambas Electronegativadades
if a < b:
  print("----------------------------------------------")
  print(f"La electronegatividad de {a} es menor a {b}")
elif a > b:
  print("----------------------------------------------")
  print(f"La electronegatividad de {a} es mayor a {b}")
else:
  print("----------------------------------------------")
  print(f"La electronegatividad de {a} es igual a {b}")
```

```python
#Ejercicio 8. Identificación del tipo de Enlace según Diferencia de Electronegatividad:

#Solicitud de la Diferencia de Electronegatividad
delta = float(input("Ingrese la diferencia de electronegatividad: "))

#Reconocimiento/Determinación del Tipo de Enlace
if delta >= 0 and delta < 0.5:
  print("---------------------------------------------")
  print("Su compuesto tiene enlace Covalente No Polar")
elif delta >= 0.5 and delta < 2:
  print("------------------------------------------")
  print("Su compuesto tiene enlace Covalente Polar")
elif delta >= 2:
  print("---------------------------------")
  print("Su compuesto tiene enlace Iónico")
else:
  print("-------------------------------")
  print("Su valor no puede ser negativo")
```

```python
#Ejercicio 9. Determinación de un Reactivo Limitante:

#Solicitud de los Moles de Ambos Reactivos
moles_a = float(input("Ingresa los moles del reactivo A: "))
moles_b = float(input("Ingresa los moles del reactivo B: "))

#Reconocimiento/Determinación del Reactivo Limitante
if moles_a < moles_b:
  print("--------------------------------")
  print(f"Su reactivo limitante es A {moles_a}")
elif moles_b < moles_a:
  print("--------------------------------")
  print(f"Su reactivo limitante es B {moles_b}")
else:
  print("-----------------------------------------")
  print(f"Su reactivo limitante es A {moles_a} y B {moles_b}")
```

```python
#Ejercicio 10. Clasificación del Rendimiento de una Reacción:

#Solicitud de Valor Teórico y Experimental
teórico = float(input("Ingrese el valor teórico: "))
experimental = float(input("Ingrese el valor experimental: "))

#Calculo de Rendimiento Porcentual
porcentaje = (experimental / teórico) * 100
print("---------------------------------------")
print(f"Su porcentaje de rendimiento es: {porcentaje}")

#Reconocimiento/Clasificación del Rendimiento
if porcentaje <= 50 and porcentaje > 0:
  print("---------------------------------------")
  print("Clasificación: Rendimiento Bajo")
elif porcentaje <= 85 and porcentaje > 50:
  print("---------------------------------------")
  print("Clasificación: Rendimiento Aceptable")
elif porcentaje < 125 and porcentaje > 85:
  print("---------------------------------------")
  print("Clasificación: Rendimiento Perfecto")
elif porcentaje < 150 and porcentaje >= 125:
  print("---------------------------------------")
  print("Clasificación: Rendimiento Elevado")
elif porcentaje <= 200 and porcentaje >= 150:
  print("---------------------------------------")
  print("Clasificación: Rendimiento Muy Elevado")
else:
  print("---------------------------------------")
  print("Clasificación: Anomalía")
```
