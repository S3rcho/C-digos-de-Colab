```python
#Ejercicio 1. Listado de Elementos y Símbolo:

#Lista de Elementos y Símbolo
elementos = ["Hidrógeno(H)", "Helio(He)", "Litio(Li)", "Berilio(Be)", "Boro(B)", "Carbono(C)", "Nitrógeno(N)", "Oxígeno(O)", "Flúor(F)", "Neón(Ne)"]

#Mensaje de Ubicación para el "Lector"
print("Listado de algunos Elementos:")
print("----------------------")

#Establecimiento del Ciclo y sus Características
for n in elementos:
  print(f"Elemento: {n}")
  print("----------------------")
```

```python
#Ejercicio 2. Listado de Compuestos y Masas Molares:

#Lista de Compuestos y Masa Molar
compuestos = {"H2O": 18.015, "CO2": 44.01, "NaCl": 58.44, "NH3": 17.031}

#Mensaje de Ubicación para el "Lector"
print("Listado de Compuestos/Masas Molares:")

#Establecimiento del Ciclo y sus Características
for formula, masa_molar in compuestos.items():
  print("----------------------------------------------")
  print(f"Información del Compuesto: {formula} - {masa_molar} g/mol")
```

```python
#Ejercicio 3. Conversión de Temperaturas en Celsius (°C) a Kelvin (de 10 en 10):

#Mensaje de Ubicación para el "Lector"
print("Listado de Temperaturas en Celsius(°C) a Kelvin(K):")

#Establecimiento del Ciclo y sus Características
for temperatura in range(0, 101, 10):
  conversion = temperatura + 273.15
  print("------------------------")
  print(f"{temperatura}°C equivale a {conversion}K")
```

```python
#Ejercicio 4. Cálculo de Moles de Agua (H2O) según Masa Molar Total:

#Lista de Masas Molares y Valor Divisor
masas_molares = [18.015, 36.03, 45, 90]
masa_agua = 18.015

#Mensaje de Ubicación para el "Lector"
print("Listado de Moles de Agua(H2O)según Masa Molar:")

#Establecimiento del Ciclo y sus Características
for masa in masas_molares:
  moles = masa / masa_agua
  print("------------------------")
  print(f"Masa Molar: {masa} g/mol")
  print(f"-Moles de Agua: {moles:.2f} mol")
```

```python
#Ejercicio 5. Cálculo del Promedio de 5 Valores de pH:

#Lista de Valores-pH y Variable "Sumando"
valores_pH = [6.8, 7.1, 6.9, 7.0, 7.2]
sumatoria = 0

#Mensaje de Ubicación para el "Lector"
print("Promedio de Valores de pH:")
print(valores_pH)

#Establecimiento del Ciclo y sus Características
for pH in valores_pH:
  sumatoria += pH
  print("---------------------------")
  print(f"Progreso de Suma: {sumatoria:.2f} pH")
promedio = sumatoria / len(valores_pH) #Número de Valores en Conjunto
print("---------------------------")
print(f"Promedio Final: {promedio:.2f} pH")
```

```python
#Ejercicio 6. Clasificación de Valores de pH según Ácido, Básico o Neutro:

#Lista de Valores-pH
valores_pH = [2, 5.5, 7, 8.3, 12]

#Mensaje de Ubicación para el "Lector"
print("Clasificación de Valores de pH:")
print(valores_pH)

#Establecimiento del Ciclo y sus Características
for clasificacion in valores_pH:
  if clasificacion < 7 and clasificacion >= 0:
    print("------------------------")
    print(f"Valor de pH: {clasificacion}: Ácido")
  elif clasificacion > 7 and clasificacion <= 14:
    print("------------------------")
    print(f"Valor de pH: {clasificacion}: Básico")
  else:
    print("------------------------")
    print(f"Valor de pH: {clasificacion}: Neutro")
```

```python
#Ejercicio 7. Sumatoria de un Listado de Valores de Masa:

#Lista de Valores de Masas y Variable "Sumando"
masas = [12.5, 8.3, 15.2, 20.0, 6.5]
sumatoria = 0

#Mensaje de Ubicación para el "Lector"
print("Sumatoria de Valores de Masas:")
print(masas)

#Establecimiento del Ciclo y sus Características
for gramos in masas:
  sumatoria += gramos

  if gramos == 6.5:
    break #Para Evitar Mensajes Innecesarios

  print("--------------------------")
  print(f"Progreso de Suma: {sumatoria:.2f} g")
print("--------------------------")
print(f"Sumatoria Final: {sumatoria:.2f} g")
```

```python
#Ejercicio 8.

#Establecimiento de Constantes
R = 0.082057
V = 10
n = 1

#Mensaje de Ubicación para el "Lector"
print("Cálculo de Presiones con Temperaturas Variables:")
print(f"Contantes: R/{R}, V/{V}, n/{n}")

#Establecimiento del Ciclo y sus Características
for T in range(300, 501, 50):
  P = (n * R  * T) / V
  print("--------------------------")
  print(f"Con Temperatura: {T} K")
  print(f"Se tiene Presión: {P:.4f} atm")
```

```python
#Ejercicio 9.Conteo y Clasificación de Valores de pH Ácidos:

#Lista de Valores-pH y Variable "Sumando"
valores_pH = [2.5, 7, 9.1, 4.3, 6.8, 11]
sumatoria = 0

#Mensaje de Ubicación para el "Lector"
print("Cantidad de Valores-pH Ácidos:")
print(valores_pH)

#Establecimiento del Ciclo y sus Características
for clasificacion in valores_pH:
  if clasificacion < 7 and clasificacion >= 0:
    sumatoria += 1
    print("------------------------------")
    print(f"Valor de pH: {clasificacion}; Ácido")
    print(f"Cantidad de Ácidos: {sumatoria}")
  else:
    print("------------------------------")
    print(f"Valor de pH: {clasificacion}; No es Ácido")
print("------------------------------")
print(f"Cantidad de Valores Ácido Final: {sumatoria}")
```

```python
#Ejercicio 10. Reducción en la Concentración de una Disolución:

#Establecimiento de Variables
concentracion = 2

#Mensaje de Ubicación para el "Lector"
print(f"Reducción de Concentración ({concentracion}mol/L) según se Duplica el Volumen:")

#Establecimiento del Ciclo y sus Características
for n in range(1, 6):
  concentracion /= 2
  print("----------------------------------------------------------")
  print(f"Concentración al Duplicar el Volumen({n}° vez): {concentracion:.4f}mol/L")
```
