```python
#Tarea 4; Sergio Soto:
#Importar "folium" para Mapas:
import folium

#Importar MiniMap:
from folium.plugins import MiniMap

#Coordenadas de Heredia:
latitud_heredia= 10.000024
longitud_heredia= -84.1165

#Coordenadas de Cartago:
latitud_cartago= 9.8644
longitud_cartago= -83.9194

#Coordenadas de San Jose:
latitud_sanjose= 9.9281
longitud_sanjose= -84.0907


#Crear el mapa con centro de mapa y nival de acercamiento
my_map = folium.Map(location= [latitud_sanjose, longitud_sanjose], zoom_start=10,
                    titles= "OpenStreeMap")

#Texto de aparición por "Click" en cada Provincia:
popuptext_heredia= "<b>Heredia, Costa Rica</b>"
popuptext_cartago= "<b>Cartago, Costa Rica</b>"
popuptext_sanjose= "<b>San Jose, Costa Rica</b>"

#Agregar marcador de Heredia:
folium.Marker(
    [latitud_heredia, longitud_heredia],
    popup=popuptext_heredia,
    tooltip= "Heredia",
    icon=folium.Icon(
        color= "orange",
        icon= "info-sign",
        prefix= "fa")
    ).add_to(my_map)

#Agregar marcador de Cartago:
folium.Marker(
    [latitud_cartago, longitud_cartago],
    popup=popuptext_cartago,
    tooltip= "Cartago",
    icon=folium.Icon(
        color= "blue",
        icon= "info-sign",
        prefix= "fa")
    ).add_to(my_map)

#Agregar marcador de San Jose:
folium.Marker(
    [latitud_sanjose, longitud_sanjose],
    popup=popuptext_sanjose,
    tooltip= "San Jose",
    icon=folium.Icon(
        color= "green",
        icon= "info-sign",
        prefix= "fa")
    ).add_to(my_map)


#Agregar circulo de Heredia:
folium.Circle(
    radius=1000,
    location= [latitud_heredia, longitud_heredia],
    popup=popuptext_heredia,
    color= "black",
    fill=True,
    fill_color= "orange",
    fill_opacity= 0.3,
    tooltip= "Zona Heredia"
).add_to(my_map)

#Agregar circulo de Cartago:
folium.Circle(
    radius=1000,
    location= [latitud_cartago, longitud_cartago],
    popup=popuptext_cartago,
    color= "black",
    fill=True,
    fill_color= "blue",
    fill_opacity= 0.3,
    tooltip= "Zona Cartago"
).add_to(my_map)

#Agregar circulo de San Jose:
folium.Circle(
    radius=10000,
    location= [latitud_sanjose, longitud_sanjose],
    popup=popuptext_sanjose,
    color= "black",
    fill=True,
    fill_color= "green",
    fill_opacity= 0.3,
    tooltip= "Zona San Jose"
).add_to(my_map)

#Agregar MiniMapa:
minimap=MiniMap(
    title_layer= "OpenStreetMap",
    position= "bottomright")
my_map.add_child(minimap)

#Guardar el Mapa:
my_map.save("Provincias.html")

#Mostrar el Mapa:
my_map
```
