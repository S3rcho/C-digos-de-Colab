```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
#Leer el archivo.csv.
df = pd.read_csv ("/content/School-Age-Digital-Connectivity - Primary.csv")

#Crear una tabla de contigencia (cuenta las frecuencias de combinacion de income Group y Residence urban).
contingency_table = pd.crosstab(df["Income Group"], df["Region"])
df.head()
#Crear mapa de calor usando seaborn. La biblioteca trae la funcion "headmap", la cial nos permite modificar colores.
#annot: Este parametro realiza anotasciones en la celda del mapa; cmap permite personalizar el mapa.

#Crear el mapa de calor:
plt.figure(figsize=(10, 7))
sns.heatmap(contingency_table, annot=True, cmap="PiYG")
plt.title("Contingency Tabla: Income Group vs Region")
plt.show()

#Mostrar el grafico:
plt.title("Contingency Tabla: Income Group vs Region")
plt.show()
```
