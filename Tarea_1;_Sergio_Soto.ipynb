```python
#Primer Código: Operaciones de Suma, Resta y Multiplicación

#Solicitud de variables
num1 = float(input("Ingrese primer número: "))
num2 = float(input("Ingrese segundo número: "))
#Operaciones respectivas con variables definidas
suma = num1+num2
resta = num1-num2
multiplicacion = num1*num2
#Resultado de las operaciones en pantalla
print("---------------------------------------")
print("La suma entre las variables es:", suma)
print("La resta entre las variables es:", resta)
print(f"La multiplicación entre y las variables es:", multiplicacion)
```

```python
#Segundo Código: Cálculo de Área y Perímetro de un Rectángulo

#Solicitud de Variables
base = float(input("Ingrese valor de base del rectángulo: "))
altura = float(input("Ingrese valor de altura del rectángulo: "))
#Operaciones para Área y Perímetro del Rectángulo
área = base*altura
perímetro = 2*(base+altura)
#Resultados de las Operaciones en Pantalla
print("------------------------------------------")
print("El área del rectángulo es:", área)
print("El perímetro del rectángulo es:", perímetro)
```

```python
#Tercer Código: Solicitud de Precio y Aplicación de Descuento %15

#Solicitud del Precio
precio_original = float(input("Ingrese el precio original del producto: "))
#Operación del Descuento del 15%
descuento = precio_original*15/100
precio_final = precio_original-descuento
#Resultado de la Aplicación del Descuento
print("------------------------------------------")
print("El precio final del producto es:", precio_final)
```

```python
#Cuarto Código: Conversión de Horas a Minutos y Segundos

#Solicitud de Horas
horas = float(input("Ingrese la cantidad de horas: "))
#Operaciones para Conversión de Horas a Minutos y Segundos
minutos = horas*60
segundos = horas*3600
#Resultados de las Conversiones
print("------------------------------------------------------------")
print("El equivalente en minutos de las horas ingresadas es:", minutos)
print("El equivalente en segundos de las horas ingresadas es:", segundos)
```

```python
#Quinto Código: Información Personal y Cálculo de Salario

#Solicitud de Información y Datos del Salario
nombre = input("Ingrese se nombre: ")
horas_trabajadas = float(input("Ingrese las horas trabajadas: "))
valor_horas = float(input("Ingrese el valor de sus horas: "))
#Cálculo del Salario Horas/Valor
salario = horas_trabajadas*valor_horas
#Resultados de Información Personal y Salario
print("--------------------------------")
print("Su nombres es:", nombre)
print("Su salario es:", salario)
```
