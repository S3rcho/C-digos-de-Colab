```python
import pandas as pd

# 1. Definir los datos
datos = {
    'nombre': ['Brandon', 'Sergio', 'Jara', 'Makelele', 'Nemo'],
    'edad': [19, 18, 18, 32, 20],
    'ciudad': ['San José', 'Alajuela', 'Heredia', 'Limon', 'Guanacaste'],
    'seccion': ['A', 'A', 'A', 'B', 'B']
}

df = pd.DataFrame(datos)

# 2. Imprimir el DataFrame
print("INFORMACIÓN DEL DATAFRAME")
print(df)

# 3. Buscar una columna
def buscar_columna(df, trabajo):
    if trabajo in df.columns:
        return df[trabajo]
    return "La columna no existe."

print("\nTODAS LAS EDADES")
print(buscar_columna(df, 'edad'))

# 4. Buscar un valor y devolver sus índices
def buscar_dato(df, columna, valor):
    if columna not in df.columns:
        raise ValueError("La columna no existe.")
    return df.index[df[columna] == valor].tolist()

# 5 y 6. Buscar la sección A
print("\nPERSONAS DE LA SECCIÓN A")
indices = buscar_dato(df, 'seccion', 'A')
print("Índices:", indices)
print(df.loc[indices])

# 7. Probar otras búsquedas
print("\nPERSONAS DE ALAJUELA")
print(df.loc[buscar_dato(df, 'ciudad', 'Limon')])

print("\nPERSONAS DE 20 AÑOS")
print(df.loc[buscar_dato(df, 'edad', 20, "ciudad", "Limon")])

print("\nBUSCAR A ANA")
print(df.loc[buscar_dato(df, 'nombre', 'Nemo')])

# 8. Buscar dos condiciones simultáneamente
def buscar_dos_datos(df, columna1, valor1, columna2, valor2):
    condicion = (
        (df[columna1] == valor1) &
        (df[columna2] == valor2)
    )
    return df.index[condicion].tolist()

print("\nPERSONAS DE ALAJUELA CON 20 AÑOS")
indices = buscar_dos_datos(
    df, 'ciudad', 'Alajuela', 'edad', 20
)
print("Índices:", indices)
print(df.loc[indices])
```

```python
import pandas as pd

# 1. Definir los datos
datos = {
    'nombre': ['Brandon', 'Sergio', 'Jara', 'Makelele', 'Nemo'],
    'edad': [19, 18, 18, 32, 20],
    'ciudad': ['San José', 'Alajuela', 'Heredia', 'Limon', 'Heredia'],
    'seccion': ['A', 'A', 'A', 'B', 'B']
}

df = pd.DataFrame(datos)

# 2. Imprimir el DataFrame
print("---------------------------------------")
print("Información del DataFrame")
print(df)

# 3. Buscar una columna
def buscar_columna(df, columna):
    if columna in df.columns:
        return df[columna]
    return "La columna no existe."

print("---------------------------------------")
print("Todas las Edades: ")
print(buscar_columna(df, 'edad'))

# 4. Buscar un valor y devolver sus índices
def buscar_dato(df, columna, valor):
    if columna not in df.columns:
        raise ValueError("La columna no existe.")
    return df.index[df[columna] == valor].tolist()

# 5 y 6. Buscar la sección A
print("---------------------------------------")
print("Personas de la Sección A")
indices = buscar_dato(df, 'seccion', 'A')
print("Índices:", indices)
print(df.loc[indices])

# 7. Probar otras búsquedas
print("---------------------------------------")
print("Personas de Limon")
print(df.loc[buscar_dato(df, 'ciudad', 'Limon')])

print("---------------------------------------")
print("Personas entre 20 a 40 años")
print(df.loc[(df["edad"] >= 20) & (df["edad"] <= 40)])

print("---------------------------------------")
print("Personas de 32 años")
print(df.loc[buscar_dato(df, 'edad', 32)])

print("---------------------------------------")
print("Buscando a Nemo")
print(df.loc[buscar_dato(df, 'nombre', 'Nemo')])

# 8. Buscar dos condiciones simultáneamente
print("---------------------------------------")
print("Personas de Heredia con 20 años")
indices = buscar_dos_datos(df, 'ciudad', 'Heredia', 'edad', 20)
print("Índices:", indices)
print(df.loc[indices])
```
