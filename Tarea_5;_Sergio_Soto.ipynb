```python
#Inciso 1. Importación de libreria "Pandas" y creación de Tabla/"DataFrame" (columnas de datos):
import pandas as pd
  #Asignación de la Tabla/DataFrame "mascota" con sus respectivos datos; cambiando nombres respecto al documento por estetica propia:
mascota = pd.DataFrame(
    data= {'---Name---': ['Minino', 'Milo', 'Snoppy', 'Blanco', 'LittlePony'],
           '-Type-': ['Gato', 'Perro', 'Perro', 'Perro', 'Caballo']},

  #Asignación de valores personalizados al índice del Tabla/DataFrame:
    index=[71, 42, 11, 98, 42])

  #Asignación de la Tabla/DataFrame "visitas" con sus respectivos datos; cambiando nombres respecto al documento por estetica propia:
visitas = pd.DataFrame(
    data= {'Pet_ID': [42, 31, 71, 42, 98, 42],
           '---Fecha---': ['2019-03-15', '2019-03-15', '2019-04-05', '2019-04-06', '2019-04-12', '2019-04-12'],
           'Costo': [2000, 5000, 1000, 5000, 6000, 10000]})


  #Inciso 2. Demostración/Impresión de las Tablas/DataFrame "mascota" y "visitas" de la libreria:
print("--DataFrame: Mascotas--")#Valor estetico propio
print(mascota)
print("--------------------------------------")#Valor estetico propio
print("-----DataFrame: Visitas-----")#Valor estetico propio
print(visitas)
```

```python
#Inciso 3. Nombramiento de las columnas índice de las Tablas/DataFrame:
mascota.index.rename('-Pet_ID', inplace=True)
visitas.index.rename('-Visit_ID-', inplace=True)

  #Demostración/Impresión de las Tablas/DataFrame con variaciones hechas:
print("-----DataFrame: Mascotas-----")#Valor estetico propio
print(mascota)
print("--------------------------------------")
print("----------DataFrame: Visitas----------")#Valor estetico propio
print(visitas)


#Inciso 4. El "Inciso 3" permite un ordenamiento visual de la columna índice para lograr una mejor orientación del lector (más explicación en WORD)
```

```python
#Inciso 5. Filtrado de información sobre visita por su Pet_ID:
  #(Se podria llegar a hacer por solicitud personal de ID, pero no es solicitado en el laboratorio)
print("----------Filtro por Visita:----------")#Valor estetico propio

  #Código que filtra información de visita cuando Pet_ID = 71:
print(visitas.query('Pet_ID == 71'))

#Observación: El cambio de nombre en el índice no se mantuvo con el filtrado de visita
```

```python
#Inciso 6. Identificación de Maximo, Minimo y SumaTotal de los costos de las visitas:
  #Código que identifica una única columna y toma sus valores para "investigar" la función indicada al final:
sumatoria_costo = visitas['Costo'].sum()#Suma de Valores en columna
min_costo = visitas['Costo'].min()#Valor Minimo de columna
max_costo = visitas['Costo'].max()#Valor Maximo de columna

  #Demostración/Impresión de los valores optenidos tras sumatoria o filtrado de datos:
print(f"Suma Total de Costos: {sumatoria_costo}")
print(f"Costo Minimo de una Visita: {min_costo}")
print(f"Costo Máximo de una Visita: {max_costo}")
```

```python
#Inciso 7. Reorganizamiento de la Tabla/DataFrame, de forma descendente, según valores de Costo:
  #Código para organizar valores de Costo y luego demostrarlos/imprimirlos de forma descendente cancelando la forma ascendente:
print(visitas.sort_values(by='Costo', ascending=False))

#Observación: El cambio de nombre en el índice no se mantuvo con la reorganización de los valores
```

```python
#Inciso 8. Combinación/Fusión de ambas Tablas/DataFrame, siendo "mascota" con los datos de "visitas":
  #Código para combinar Tabla/DataFrame "mascota" con cada inciso en "visitas":
tienda = pd.merge(mascota, visitas,how='cross')

  #Demostración/Impresión de la "tienda" creada con ambas Tablas/DataFrame:
print("------Tabla Tienda (Resultado del merge)------")#Valor estetico propio
print(tienda)
```
