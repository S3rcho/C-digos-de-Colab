```python
#Importación de Libreria (Necesaria para Graficar)
import matplotlib.pyplot as plt

#Lista de Fechas y Temperaturas (Determinación de Variable con Lista)
dias = ["Lunes", "Martes", "Miércoles", "Jueves", "Viernes"]
temperaturas = [18, 22, 31, 27, 16]

#Rango "for" para Clasificar Temperaturas
for n in range(len(dias)):
    temp = temperaturas[n]
    if temp >= 30:
        clasificacion = "Caluroso"
    elif temp >= 20:
        clasificacion = "Templado"
    else:
        clasificacion = "Frío"
    #Demostración
    print(f"{dias[n]}: {temp}°C == {clasificacion}")
    print("=========================")

#Creación de Gráfica
plt.bar(dias, temperaturas)
plt.title("Temperaturas Diariamente")
plt.xlabel("Días")
plt.ylabel("Temperatura (°C)")

#Demostración de la Gráfica
plt.show()
```

```python
#Importación de Libreria (Necesaria para Graficar)
import matplotlib.pyplot as plt

#Lista de Calificaciones (Determinación de Variable con Lista)
calificaciones = [85, 90, 75, 70, 65, 50, 40]
#Variables Iniciales
aprobados = 0
reprobados = 0

#Rango "for" para Clasificar Notas
for nota in calificaciones:
    if nota >= 70:
        aprobados += 1
    else:
        reprobados += 1
#Demostración
print("Aprobados: ", aprobados)
print("==============")
print("Reprobados:" ,reprobados)
print("==============")

#Determinación de Listas para Gráfica
categorias = ["Aprobados", "Reprobados"]
cantidades = [aprobados, reprobados]
#Creación de Gráfica
plt.bar(categorias, cantidades)
plt.title("Cantidad de Estudiantes Aprobados o Reprobados")
plt.ylabel("Cantidad Estudiantes")

#Demostración de la Gráfica
plt.show()
```

```python
#Importación de Libreria (Necesaria para Graficar)
import matplotlib.pyplot as plt

#Lista de Estudiantes y Notas (Determinación de Variable con Lista)
estudiantes = ["Emanuel", "Brandon", "Kendall", "Sergio", "Jara"]
notas = [80, 69, 91, 67, 79]

#Rango "for" para Clasificar Notas
for n in range(len(notas)):
    if notas[n] >= 90:
        clasificacion = "Excelente"
    elif notas[n] >= 70:
        clasificacion = "Aprobado"
    else:
        clasificacion = "Reprobado"
    print(f"{estudiantes[n]}: {notas[n]} - {clasificacion}")
    print("=======================")

#Creación de Gráfica
plt.bar(estudiantes, notas)
plt.title("Promedio Calificaciones Estudiantes Química")
plt.xlabel("Estudiantes")
plt.ylabel("Promedio Calificaciones")

#Demostración de la Gráfica
plt.show()
```

```python
#Importación de Libreria (Necesaria para Graficar)
import matplotlib.pyplot as plt

#Solicitud de Valor Numerico al Usuario
num = int(input("Ingrese un número para la tabla de multiplicar: "))

multiplicadores = list(range(1, 11))
resultados = []

# Generar tabla con un ciclo for
for n in multiplicadores:
    resultados.append(num * n)

#Creación de Gráfica
plt.plot(multiplicadores, resultados, marker='o')
plt.title("Tabla de multiplicar")
plt.xlabel("Multiplicador")
plt.ylabel("Resultado")

#Demostración de la Gráfica
plt.show()
```

```python
#Importación de Libreria (Necesaria para Graficar)
import matplotlib.pyplot as plt

#Lista de Fechas y Temperaturas (Determinación de Variable con Lista)
dias = ["Lun", "Mar", "Mié", "Jue", "Vie", "Sáb", "Dom"]
temperaturas = [18, 25, 32, 28, 16, 30, 22]

#Rango "for" para Clasificar Temperaturas
for n in range(len(dias)):
    temp = temperaturas[n]
    if temp >= 30:
        clasificacion = "Calurosa"
    elif temp >= 20:
        clasificacion = "Agradable"
    else:
        clasificacion = "Fría"
    print(f"{dias[n]}: {temp}°C - {clasificacion}")
    print("======================")

#Creación de Gráfica
plt.plot(dias, temperaturas, marker='o')
plt.title("Temperaturas Diariamente")
plt.xlabel("Fechas")
plt.ylabel("Temperatura (°C)")

#Demostración de la Gráfica
plt.show()
```

```python
#Importación de Libreria (Necesaria para Graficar)
import matplotlib.pyplot as plt

#Lista de Fechas y Ventas (Determinación de Variable con Lista)
dias = ["Lunes", "Martes", "Miércoles", "Jueves", "Viernes"]
ganancias = [120000, 45000, 80000, 150000, 30000]

#Rango "for" para Clasificar Ventas/Ganancias
for n in range(len(dias)):
    venta = ganancias[n]
    if venta >= 100000:
        clasificacion = "Venta Alta"
    elif venta >= 50000:
        clasificacion = "Venta Media"
    else:
        clasificacion = "Venta Baja"
    print(f"{dias[n]}: ₡{venta} == {clasificacion}")
    print("================================")

#Creación de Gráfica
plt.bar(dias, ventas)
plt.title("Ganancias Diarias")
plt.xlabel("Fechas")
plt.ylabel("Ganancias (Colones)")

#Demostración de la Gráfica
plt.show()
```

```python
#Importación de Libreria (Necesaria para Graficar)
import matplotlib.pyplot as plt

#Lista de Números Enteros (Determinación de Variable con Lista)
numeros = [10, -10, 0, 5, -5, 0, 20, -20, 0, 100]
#Variables Iniciales
positivos = 0
negativos = 0
ceros = 0

#Rango "for" para Clasificar Números Enteros
for n in numeros:
    if n > 0:
        positivos += 1
    elif n < 0:
        negativos += 1
    else:
        ceros += 1
#Demostración de Cantidades de Positivos, Negativos y Ceros
print(f"Valores Positivos: {positivos}")
print("====================")
print(f"Valores Negativos: {negativos}")
print("====================")
print(f"Valores Ceros: {ceros}")
print("====================")

#Determinación de Listas para Gráfica
clasificacion = ["Positivos", "Negativos", "Ceros"]
cantidades = [positivos, negativos, ceros]
#Creación de Gráfica
plt.bar(clasificacion, cantidades)
plt.title("Valoración de Números Enteros")
plt.xlabel("Clasificación")
plt.ylabel("Cantidad")

#Demostración de la Gráfica
plt.show()
```
